### Пример 1.8.
В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

* Пусть А - варианты распределения призов, если по каждой номинации установлены различные призы

$ \overline{A}_{10}^5 = 10^5$

```python
n = 10
k = 5
```

```python
# а)
a = n**k
a
```

$ \overline{C}_{10 + 5 - 1}^5 = C_{14}^5 = \frac{14!}{5!(14-5)!}$

* Пусть B - варианты распределения призов, если по каждой номинации установлены одинаковые призы

```python
# б)
b = sc.comb(n+k-1, k)
b
```

### Пример 1.9.
Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

$P_{7} = \frac{7!}{3!2!2!}$

```python
a = sc.factorial(7)/(sc.factorial(3)*sc.factorial(2)*sc.factorial(2))
a
```

### Пример 1.10.
Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Пусть событие A - получилось слово "ТОР"

$ P(A) = \frac{1}{A_{6}^3}$

```python
# а)
alph = 'ТЕИЯРО'
a = round(1/len(list(permutations(alph, 3))), 5)
a
```

* Пусть событие B - получилось слово "ТЕОРИЯ"

$ P(B) = \frac{1}{P_{6}}$ 

```python
# б)
b = round(1/len(list(permutations(alph, 6))), 5)
b
```

### Пример 1.11.
Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

* Пусть событие A - получилось слово АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С

### Пример 1.16.
За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

* Пусть событие A - никакие два лица одного пола не сядут рядом

$ P(A) = \frac{2}{P_{10}(5, 5)}$

```python
# а)
a = round(2/(sc.factorial(10)/(sc.factorial(5)*sc.factorial(5))), 5)
a
```

* Пусть событие B - супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар

$ P(B) = \frac{5!\cdot2^5}{10!}$

```python
# б)
b = round(sc.factorial(5)*2**5/sc.factorial(10), 5)
b
```

### Пример 1.16а.
В аудитории m = 25 студентов. Найти вероятность того, что хотя бы у двух студентов дни рождения совпадают. При каком числе m студентов вероятность того же события не меньше чем 0,95? (Полагаем равновозможность рождений в любой день года.)

* Пусть событие A - хотя бы у двух студентов дни рождения совпадают

$P(A) = 1 - \overline{P}(A) $

$\overline{P}(A) = \frac{A_{365}^{25}}{\overline{A}_{365}^{25}}$

```python
m = 25
n = 365
inv_a = sc.perm(n, m)/n**m
print(round(inv_a, 5))
a = 1 - inv_a
round(a, 5)
```

```python
for i in range(1, 2*m):
    inv_a = sc.perm(n, i)/n**i
    a = 1 - inv_a
    if a >= 0.95:
        print(a)
        print(i)
        break
```

### Пример 1.17.
В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

* Пусть событие A - пассажиры попали в 2 купе

$ P(A) = \frac{C_{9}^2 C_{4}^4 C_{4}^3 \cdot 2}{C_{36}^7}$

```python
# а)
a = sc.comb(9, 2)*sc.comb(4, 4)*sc.comb(4, 3)*2/sc.comb(36, 7)
round(a, 7)
```

* Пусть событие B - пассажиры попали в 7 купе

$ P(B) = \frac{C_{9}^7 \cdot 4^7}{C_{36}^7}$

```python
# б)
b = sc.comb(9, 7)*4**7/sc.comb(36, 7)
round(b, 5)
```

* Пусть событие C - пассажиры попали в 3 купе

$ P(С) = \frac{C_{9}^3
(6 \cdot C_{4}^4 C_{4}^2 C_{4}^1 +
 3 \cdot C_{4}^3 C_{4}^3 C_{4}^1 +
 3 \cdot C_{4}^2 C_{4}^2 C_{4}^3)}{C_{36}^7}$

```python
# в)
c = sc.comb(9, 3)* \
    (6*sc.comb(4, 4)*sc.comb(4, 2)*sc.comb(4, 1) + \
     3*sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1) + \
     3*sc.comb(4, 2)*sc.comb(4, 2)*sc.comb(4, 3)) / \
     sc.comb(36, 7)
c
```
