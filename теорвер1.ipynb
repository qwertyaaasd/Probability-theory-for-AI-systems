## 1.16a

$P = 1 - \overline{P} $

$\overline{P} = \frac{A_{365}^{25}}{\overline{A}_{365}^{25}}$

```python
m = 25
n = 365
inv_a = sc.perm(n, m)/n**m
print(round(inv_a, 5))
a = 1 - inv_a
round(a, 5)
```

## 1.17
