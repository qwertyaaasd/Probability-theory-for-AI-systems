### 1.38.
Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

* Пусть событие A - книги стоят слева направо в порядке нумерации томов

$ P(A) = \frac{1}{P_{5}}$

```python
a = round(1/sc.factorial(5), 5)
a
```

### 1.43.
Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не на- чинается с цифры ноль.

* Пусть событие A - все цифры различны

$P(A) = \frac{9 \cdot A_{9}^5}{9 \cdot \overline{A}_{10}^4}$

```python
# а)
a = 9*sc.perm(9, 4)/(9*10**4)
a
```

* Пусть событие B - все цифры одинаковые

$ P(B) = \frac{9}{9 \cdot \overline{A}_{10}^4}$

```python
# б)
b = 1/10**4
b
```

* Пусть событие C - все цифры нечетные

$ P(C) = \frac{\overline{A}_{5}^5}{9 \cdot \overline{A}_{10}^4}$

```python
# в)
c = round(5**5/(9*10**4), 4)
c
```

### 1.48.
В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

* Пусть событие A - выпало 11 очков

```python
alph = '123456'
a = list(product(alph, repeat=3))
```

```python
# a)
cnt = 0
for el in a:
    summ = 0
    for i in el:
        summ += int(i)
    if summ == 11:
        cnt += 1

p_a = cnt/6**3
p_a
```

* Пусть событие B - выигрыш

```python
# б)
cnt = 0
for el in a:
    summ = 0
    for i in el:
        summ += int(i)
    if summ >= 11:
        cnt += 1

p_b = cnt/6**3
p_b
```
